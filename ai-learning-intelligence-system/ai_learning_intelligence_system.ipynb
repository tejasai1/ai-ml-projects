```python
from sentence_transformers import SentenceTransformer
import time

# ── Load embedding model ──
print("Loading embedding model...")
print("(This downloads ~90MB the first time — takes 30-60 seconds)")

embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
print("Embedding model loaded!")

# ── Generate embeddings for all documents ──
print("\nGenerating embeddings for all 19 documents...")
start = time.time()

# Extract just the content text for embedding
texts = []
for doc in knowledge_base:
    # Combine title + content for richer embedding
    combined = f"{doc['title']}. {doc['content'].strip()}"
    texts.append(combined)

# Generate all embeddings at once (batch processing)
embeddings = embedding_model.encode(
    texts,
    show_progress_bar=True,
    batch_size=8
)

elapsed = time.time() - start

print(f"\nEmbeddings generated in {elapsed:.1f} seconds")
print(f"Total embeddings: {len(embeddings)}")
print(f"Embedding dimensions: {embeddings.shape[1]}")
print(f"Each document is now a vector of {embeddings.shape[1]} numbers")

# Show sample embedding for first document
print(f"\nSample — first 5 numbers of '{knowledge_base[0]['title']}':")
print(embeddings[0][:5].tolist())
```

```python
import chromadb

# ── Create ChromaDB client and collection ──
chroma_client = chromadb.Client()

# Delete collection if it already exists (clean start)
try:
    chroma_client.delete_collection("ai_learning_kb")
    print("Existing collection deleted — starting fresh")
except:
    pass

collection = chroma_client.create_collection(
    name="ai_learning_kb",
    metadata={"description": "AI/ML learning knowledge base for RAG system"}
)
print("ChromaDB collection created: ai_learning_kb")

# ── Store all documents with embeddings and metadata ──
print("\nStoring 19 documents in ChromaDB...")

collection.add(
    ids=[doc['id'] for doc in knowledge_base],
    embeddings=embeddings.tolist(),
    documents=[doc['content'].strip() for doc in knowledge_base],
    metadatas=[{
        "title": doc['title'],
        "topic": doc['topic'],
        "id": doc['id']
    } for doc in knowledge_base]
)

# ── Verify storage ──
count = collection.count()
print(f"Documents stored: {count}")
print("\nSample — first 3 stored documents:")
sample = collection.get(
    ids=['py_001', 'ann_001', 'rag_001'],
    include=['metadatas']
)
for meta in sample['metadatas']:
    print(f"  ✓ [{meta['topic']}] {meta['title']}")

print("\nChromaDB storage complete!")
print("Your knowledge base is now searchable by meaning.")
```

```python
def retrieve(query, n_results=3, min_confidence=0.3):
    """
    Retrieve most relevant documents for a query.

    Args:
        query: natural language question
        n_results: number of documents to retrieve (top-k)
        min_confidence: minimum confidence threshold (0-1)

    Returns:
        dict with retrieved docs, scores, and confidence assessment
    """

    # Step 1 — embed the query
    query_embedding = embedding_model.encode([query]).tolist()

    # Step 2 — search ChromaDB
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=n_results,
        include=['documents', 'metadatas', 'distances']
    )

    # Step 3 — convert distances to confidence scores
    # ChromaDB returns L2 distances — lower = more similar
    # Convert to 0-1 confidence: confidence = 1 / (1 + distance)
    retrieved = []
    for i in range(len(results['documents'][0])):
        distance = results['distances'][0][i]
        confidence = round(1 / (1 + distance), 3)

        retrieved.append({
            'content':    results['documents'][0][i],
            'title':      results['metadatas'][0][i]['title'],
            'topic':      results['metadatas'][0][i]['topic'],
            'id':         results['metadatas'][0][i]['id'],
            'confidence': confidence,
            'distance':   round(distance, 4)
        })

    # Step 4 — assess overall retrieval confidence
    if retrieved:
        top_confidence = retrieved[0]['confidence']
        if top_confidence >= 0.6:
            assessment = "HIGH"
        elif top_confidence >= min_confidence:
            assessment = "MEDIUM"
        else:
            assessment = "LOW — possible knowledge gap"
    else:
        assessment = "NO RESULTS"
        top_confidence = 0.0

    return {
        'query':          query,
        'results':        retrieved,
        'top_confidence': top_confidence,
        'assessment':     assessment,
        'gap_detected':   top_confidence < min_confidence
    }


# ── Test the retrieval function ──
print("Testing retrieval function...\n")

test_queries = [
    "What is RAG and how does it work?",
    "How does inter-annotator agreement work?",
    "What is a transformer in AI?",
    "How do I use ChromaDB in Python?",
    "What is quantum computing?",   # not in knowledge base — should detect gap
]

for query in test_queries:
    result = retrieve(query, n_results=2)
    print(f"Query: {query}")
    print(f"Confidence: {result['top_confidence']} [{result['assessment']}]")
    print(f"Top result: {result['results'][0]['title']} [{result['results'][0]['topic']}]")
    if result['gap_detected']:
        print("⚠️  KNOWLEDGE GAP DETECTED")
    print()
```

```python
def retrieve(query, n_results=3, min_confidence=0.44):
    """
    Retrieve most relevant documents for a query.

    Confidence thresholds (final calibration):
        >= 0.50 → HIGH   — strong answer available
        >= 0.44 → MEDIUM — partial answer available
        <  0.44 → LOW    — knowledge gap detected
    """
    query_embedding = embedding_model.encode([query]).tolist()

    results = collection.query(
        query_embeddings=query_embedding,
        n_results=n_results,
        include=['documents', 'metadatas', 'distances']
    )

    retrieved = []
    for i in range(len(results['documents'][0])):
        distance   = results['distances'][0][i]
        confidence = round(1 / (1 + distance), 3)
        retrieved.append({
            'content':    results['documents'][0][i],
            'title':      results['metadatas'][0][i]['title'],
            'topic':      results['metadatas'][0][i]['topic'],
            'id':         results['metadatas'][0][i]['id'],
            'confidence': confidence,
            'distance':   round(distance, 4)
        })

    top_confidence = retrieved[0]['confidence'] if retrieved else 0.0

    if top_confidence >= 0.50:
        assessment   = "HIGH"
        gap_detected = False
    elif top_confidence >= 0.44:
        assessment   = "MEDIUM"
        gap_detected = False
    else:
        assessment   = "LOW — knowledge gap detected"
        gap_detected = True

    return {
        'query':          query,
        'results':        retrieved,
        'top_confidence': top_confidence,
        'assessment':     assessment,
        'gap_detected':   gap_detected
    }

# ── Final validation test ──
print("="*60)
print("RETRIEVAL SYSTEM — FINAL VALIDATION")
print("="*60)

validation_queries = [
    # Should be HIGH or MEDIUM — in knowledge base
    ("What is RAG?",                          "IN KB"),
    ("How does inter-annotator agreement work?", "IN KB"),
    ("What is a transformer in AI?",          "IN KB"),
    ("How do I use ChromaDB?",                "IN KB"),
    ("What is LangChain?",                    "IN KB"),
    ("How does RLHF work?",                   "IN KB"),
    ("What is prompt engineering?",           "IN KB"),
    ("How does data annotation work?",        "IN KB"),
    # Should be LOW — not in knowledge base
    ("What is quantum computing?",            "NOT IN KB"),
    ("How do I invest in stocks?",            "NOT IN KB"),
    ("What is the weather in Boston?",        "NOT IN KB"),
    ("How do I cook pasta?",                  "NOT IN KB"),
]

correct = 0
total   = len(validation_queries)

for query, expected in validation_queries:
    result   = retrieve(query, n_results=1)
    in_kb    = not result['gap_detected']
    expected_in_kb = expected == "IN KB"
    is_correct     = in_kb == expected_in_kb
    correct  += 1 if is_correct else 0
    status   = "✅ CORRECT" if is_correct else "❌ WRONG"
    gap_icon = "📚" if in_kb else "⚠️ "
    print(f"{status} {gap_icon} [{result['top_confidence']}] {query}")

print(f"\nAccuracy: {correct}/{total} ({correct/total*100:.0f}%)")
print("\nRetrieval system calibrated and ready!")
```

```python
import random

def generate_quiz(topic=None, num_questions=5):
    """
    Generate a quiz from the knowledge base.
    If topic is specified, generates questions for that topic only.
    If topic is None, generates questions across all topics.
    """
    print(f"\n{'='*60}")
    if topic:
        print(f"QUIZ — Topic: {topic}")
    else:
        print(f"QUIZ — Mixed topics")
    print('='*60)

    # Step 1 — select documents to quiz on
    if topic:
        topic_docs = [doc for doc in knowledge_base
                      if doc['topic'].lower() == topic.lower()]
        if not topic_docs:
            print(f"Topic '{topic}' not found in knowledge base.")
            print(f"Available topics: {list(set(d['topic'] for d in knowledge_base))}")
            return None
        selected_docs = random.sample(
            topic_docs, min(num_questions, len(topic_docs))
        )
    else:
        selected_docs = random.sample(
            knowledge_base, min(num_questions, len(knowledge_base))
        )

    # Step 2 — build quiz generation prompt
    docs_text = ""
    for i, doc in enumerate(selected_docs, 1):
        docs_text += f"\nDocument {i} [{doc['topic']}]: {doc['title']}\n{doc['content'][:300]}...\n"

    prompt = f"""You are an expert AI/ML tutor creating a quiz for a graduate student
preparing for AI internships.

Based on the following study materials, generate exactly {num_questions} multiple choice
questions. Each question must have 4 options (A, B, C, D) and one correct answer.

Study materials:
{docs_text}

Format each question EXACTLY like this:
Q1: [question text]
A) [option]
B) [option]
C) [option]
D) [option]
Answer: [correct letter]
Explanation: [one sentence explaining why]

Generate {num_questions} questions now:"""

    # Step 3 — call Groq
    response = groq_client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=1500,
        temperature=0.7
    )

    quiz_text = response.choices[0].message.content

    # Step 4 — parse quiz into structured format
    questions = []
    lines = quiz_text.strip().split('\n')
    current_q = {}

    for line in lines:
        line = line.strip()
        if not line:
            continue
        if line.startswith('Q') and ':' in line and len(line) < 200:
            if current_q and 'question' in current_q:
                questions.append(current_q)
            current_q = {'question': line.split(':', 1)[1].strip(),
                         'options': {}, 'answer': '', 'explanation': ''}
        elif line.startswith(('A)', 'B)', 'C)', 'D)')):
            letter = line[0]
            current_q['options'][letter] = line[3:].strip()
        elif line.startswith('Answer:'):
            current_q['answer'] = line.replace('Answer:', '').strip()
        elif line.startswith('Explanation:'):
            current_q['explanation'] = line.replace('Explanation:', '').strip()

    if current_q and 'question' in current_q:
        questions.append(current_q)

    # Step 5 — display quiz interactively
    print(f"\nGenerated {len(questions)} questions. Let's begin!\n")

    score = 0
    results = []

    for i, q in enumerate(questions, 1):
        print(f"Q{i}: {q['question']}")
        for letter, option in q['options'].items():
            print(f"   {letter}) {option}")

        # Get user answer
        while True:
            user_answer = input("\nYour answer (A/B/C/D) or 'skip': ").strip().upper()
            if user_answer in ['A', 'B', 'C', 'D', 'SKIP']:
                break
            print("Please enter A, B, C, D, or 'skip'")

        if user_answer == 'SKIP':
            print(f"Skipped. Correct answer: {q['answer']}")
            print(f"Explanation: {q['explanation']}\n")
            results.append({'question': q['question'], 'status': 'skipped',
                           'correct_answer': q['answer']})
        elif user_answer == q['answer']:
            score += 1
            print(f"✅ Correct! {q['explanation']}\n")
            results.append({'question': q['question'], 'status': 'correct',
                           'user_answer': user_answer, 'correct_answer': q['answer']})
        else:
            print(f"❌ Incorrect. Correct answer: {q['answer']}")
            print(f"Explanation: {q['explanation']}\n")
            results.append({'question': q['question'], 'status': 'incorrect',
                           'user_answer': user_answer, 'correct_answer': q['answer']})

    # Step 6 — show score
    answered = len([r for r in results if r['status'] != 'skipped'])
    print(f"\n{'='*60}")
    print(f"QUIZ COMPLETE — Score: {score}/{answered} ({score/max(answered,1)*100:.0f}%)")
    print('='*60)

    if answered > 0:
        pct = score / answered * 100
        if pct >= 80:
            print("🌟 Excellent! You have strong understanding of this material.")
        elif pct >= 60:
            print("👍 Good work! Review the questions you missed.")
        else:
            print("📚 Keep studying! Focus on the topics you missed.")

    # Step 7 — identify weak topics
    weak_topics = []
    for r in results:
        if r['status'] == 'incorrect':
            weak_topics.append(r['question'][:50])

    if weak_topics:
        print(f"\n⚠️  Topics to review:")
        for t in weak_topics:
            print(f"   → {t}...")

    return {
        'score':       score,
        'total':       answered,
        'percentage':  score / max(answered, 1) * 100,
        'results':     results,
        'weak_topics': weak_topics
    }


# ── Test the quiz generator ──
print("Starting quiz on Data Annotation...")
print("Answer each question by typing A, B, C, or D\n")
quiz_result = generate_quiz(topic="Data Annotation", num_questions=3)
```
